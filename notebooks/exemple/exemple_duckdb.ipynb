# Exemple de notebook - filtres sql et récupération des données en pandas

L'objectif de ce notebook est de fournir des exemples pour pré-filtrer les données via sql avant de charger les données dans un DataFrame pandas.


```python
import pandas as pd

pd.set_option("display.max_columns", None)  # show all cols
pd.set_option("display.max_colwidth", None)  # show full width of showing cols
pd.set_option(
    "display.expand_frame_repr", False
)  # print cols side by side as it's supposed to be
```

```python
import duckdb
DUCKDB_FILE = "../../dbt_ETM/dev.duckdb"

con = duckdb.connect(database=DUCKDB_FILE, read_only=True)
```

## Filtres

Filtrer sur l'année 2024


```python
query_2024 = """
SELECT * from dev.main_raw.raw_cites
WHERE "year" = 2024
"""

cites_2024 = con.sql(query_2024)
cites_2024_df = cites_2024.df()
cites_2024_df.head(2)
```

2. Filtrer mes donées pour en récupérer une partie

```python
where_clause = """
"year" == '2022'
"""

query_2022 = f"""
SELECT
  *
FROM dev.main_raw.raw_cites
WHERE
  {where_clause}
"""

cites_2022 = con.sql(query_2022)
cites_2022_df = cites_2022.df()
cites_2022_df.head(2)
```

## Selectionner des colonnes avant d'exécuter la requête


Selectionner les colonnes avant de charger les données permets une exécution plus rapide et limite l'usage de la mémoire.


```python
query_preselected = f"""
SELECT
  "source",
  "id"
FROM
  dev.main_raw.raw_cites
WHERE
  {where_clause}
"""
preselected = con.sql(query_preselected)
preselected_df = preselected.df()
preselected_df.head(2)
```

Avoir la liste des valeurs possibles différentes pour le champ source

```python
query = """
SELECT DISTINCT "source" from dev.main_raw.raw_cites
"""

cites_source = con.sql(query)
cites_source_df = cites_source.df()
cites_source_df
```
