# Explore eurostat dataset


```python
import pandas as pd

pd.set_option("display.max_columns", None)  # show all cols
pd.set_option("display.max_colwidth", None)  # show full width of showing cols
pd.set_option(
    "display.expand_frame_repr", False
)  # print cols side by side as it's supposed to be
```

```python
import duckdb
DUCKDB_FILE = "../../dbt_ETM/dev.duckdb"

con = duckdb.connect(database=DUCKDB_FILE, read_only=True)
```

```python
query_eurostats = """
SELECT * from dev.main_raw.raw_eurostat
"""

eurostats = con.sql(query_eurostats)
df_eurostats = eurostats.df()
df_eurostats.head(3)
```

```python
print(f"Dimensions: {df_eurostats.shape[0]} lignes, {df_eurostats.shape[1]} colonnes")
```
